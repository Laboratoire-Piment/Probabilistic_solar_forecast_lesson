# Introduction to Jupyter Notebook

Jupyter Notebook is an open-source web application that allows you to create and share documents containing live code, equations, visualizations, and narrative text. Here's a basic explanation of how it works:

## 1. Cells
Notebooks are composed of cells, which can be of different types:
- **Code cells**: Where you write and execute code (e.g., Python).
- **Markdown cells**: For writing formatted text, including headings, lists, and links.
- **Raw cells**: For unformatted text.

Jupyter Notebooks are composed of different types of cells, each serving a distinct purpose. Here’s a detailed overview of the main cell types and how to create them:

### 1.1. Code Cells
- **Purpose**: Used to write and execute code in languages supported by the Jupyter kernel (e.g., Python, R).
- **Creation**:
  - To create a code cell, select "Code" from the dropdown menu in the toolbar or simply click on the plus icon to add a new cell (it defaults to code).
- **Execution**:
  - Run the code by selecting the cell and pressing `Shift + Enter`. The output will display directly below the cell.

### 1.2. Markdown Cells
- **Purpose**: Used to write formatted text, including headings, lists, links, and LaTeX equations for mathematical notation.
- **Creation**:
  - To create a Markdown cell, select "Markdown" from the dropdown menu in the toolbar or click the plus icon and then change the cell type to Markdown.
- **Execution**:
  - After writing your text, press `Shift + Enter` to render the Markdown into formatted text.

## 5. Execution
You execute code cells by selecting the cell and pressing Shift + Enter. The output (e.g., text, plots) appears directly below the cell.
