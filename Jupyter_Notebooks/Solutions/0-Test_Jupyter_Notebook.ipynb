The second code will test if you are able to use the data that will be used during the workshop. It will dowload the data and plot a graph. As previously, run the cell by clicking the "play" botton.

```python
# Import libraries
import pandas as pd
import matplotlib.pyplot as plt

# Download PV production data
df_PV = pd.read_csv('https://github.com/Laboratoire-Piment/TwInSolar_consolidated_data/raw/main/ENERPOS_PV.txt')
df_PV['datetime'] = pd.to_datetime(df_PV['datetime'])
df_PV.set_index('datetime', inplace=True)

# Plot daily PV production
df_PV_d = df_PV.resample('1d').sum()/6
plt.figure(figsize=(16,5))
ax = df_PV_d['Prod_kW'].plot()
ax.set_ylabel('Daily production [kW/h]')
ax.set_xlabel('Date')
ax.set_title('Daily production of the ENERPOS PV plant in 2021 and 2022')
plt.tight_layout()
plt.show()
```
