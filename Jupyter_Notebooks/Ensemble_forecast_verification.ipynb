<h1>Ensemble forecast</h1>

Ensemble forecast are the most common probabilistic forecast issued by Numerical Weather Prediction (NWP) systems. The aim of this Notebook is to visualize and assess the quality of ensemble forecast of solar irradiance.

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import xarray as xr
import fsspec
```

<h2>Import and read data</h2>

```python
# Dowload the required data from GitHub (https://github.com/Laboratoire-Piment/Probabilistic_solar_forecast_lesson/Data)
'''destination = "Data/EPS_cf_GHI_20220601_00.nc"
fs = fsspec.filesystem("github", org="Laboratoire-Piment", repo="Probabilistic_solar_forecast_lesson/Data")
fs.get("EPS_cf_GHI_20220601_00.nc", destination)'''
```

```python
# Read the file
ds_nwp = xr.open_dataset("../Data/EPS_cf_GHI_20220601_00.nc")
ds_nwp
```

```python
# Get GHI_nwp values for the first base_time across all steps
GHI_nwp_values = np.diff(ds_nwp['ssrd'].squeeze().values)

# Plotting
plt.figure(figsize=(10, 5))  # Set the figure size
plt.plot(GHI_nwp_values, marker='o')  # Plot the GHI_nwp values with markers
plt.title(f'GHI_nwp over horizons')  # Title of the plot
plt.xlabel('Steps')  # X-axis label
plt.ylabel('GHI_nwp')  # Y-axis label
plt.xticks(rotation=45)  # Rotate x-axis labels for better readability
plt.grid()  # Add a grid to the plot
plt.show()  # Display the plot
```
