<h2>Information on NetCDF file</h2>

This NetCDF file contains irradiance data derived from NWP. It includes multiple dimensions, coordinates, and data variables that provide comprehensive information about the irradiance measurements over time and space.

<h3>Dimensions:</h3>

* <b>latitude</b>: Latitude
* <b>longitude</b>: Longitude
* <b>time</b>: Valid time of the forecast

<h3>Coordinates:</h3>

* <b>latitude</b>: Gives the value in degrees of the location latitude
* <b>longitude</b>: Gives the value in degrees of the location longitude
* <b>time</b>: A datetime coordinate representing the valid time of the forecast

<h3>Data Variables:</h3>

* <b>ssrd</b>: Surface Shortwave Radiation Downwelling (representing the cumulative GHI in the energy domain) in $J.m^{-2}$ from ECMWF for each combination of `latitude`, `longitude` and `time`.

<h3>Attributes:</h3>
Provides additional information on the dataset used


This file structure facilitates the analysis and visualization of solar irradiance data, enabling better understanding and forecasting of solar energy patterns.

<h2>Information on NetCDF file</h2>

<h3>Dimensions:</h3>

* <b>basetime</b>: Representing the time when the forecast is done
* <b>step</b>: Steps indicating the time steps from the base time, also called horizon.

<h3>Coordinates:</h3>

* <b>basetime</b>: A datetime coordinate
* <b>step</b>: A timedelta coordinate representing the time steps after the base time.

<h3>Data Variables:</h3>

* <b>GHI_nwp</b>: Forecasted GHI ($W.m^{-2}$) from ECMWF for each combination of `basetime` and `step`.
* <b>GHI_sat</b>: Forecasted GHI ($W.m^{-2}$) derived from MSG-IO for each combination of `basetime` and `step`.
* <b>GHI_pers</b>: Forecasted GHI ($W.m^{-2}$) from smart persistence for each combination of `basetime` and `step`.
* <b>GHI_clear</b>: GHI ($W.m^{-2}$) from Copernicus Atmosphere Monitoring Service (CAMS) McClear Clear-Sky for each combination of `basetime` and `step`.
* <b>GHI_meas</b>: Measured GHI ($W.m^{-2}$) for each combination of `basetime` and `step`.
* <b>zenith</b>: Solar zenith angle (°) for each combination of `basetime` and `step`.

```python
# Select the first for basetime values
basetimes = ds_det['basetime'][0:4]

plt.rcParams.update({'font.size': 10})
plt.figure(figsize=(10, 5))

# Loop over each basetime and plot
for basetime in basetimes:
    # Calculate valid_time for each base_time
    valid_times = basetime + ds_det['step']

    # Select GHI values for the current basetime
    GHI_values = ds_det['GHI_nwp'].sel(basetime=basetime).squeeze()

    # Format base_time for the label (hours and minutes)
    label_time = basetime.dt.strftime('%Y-%m-%d %H:%M').values  # Format to include date, hours, and minutes

    # Plotting
    plt.plot(valid_times.values, GHI_values.values, marker='o', label=f'base_time: {label_time}')

# Adding labels and title
plt.title('Forecast over Valid Time for different base-times')
plt.xlabel('Valid Time')
plt.ylabel('GHI ($W.m^{-2}$)')
plt.xticks(rotation=45)
plt.legend()
plt.grid()
plt.show()
```
